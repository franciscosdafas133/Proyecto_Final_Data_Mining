## **1. Introducción y objetivo**

<a name='tema'></a>

**Tema**

**Perfiles de brecha tecnológica y conectividad digital en los locales educativos del Perú**

<a name='problema'></a>

**Problema**

En Perú, una parte importante de los locales educativos reporta acceso a internet y otra parte no, pero no se conoce cómo se diferencian estos grupos en términos de infraestructura tecnológica, cobertura territorial y disponibilidad real de servicios digitales. La principal característica que orienta este estudio es precisamente la presencia o ausencia de conexión, porque permite comparar las condiciones de conectividad junto con el equipamiento y la cobertura del entorno. Estas fuentes no miden directamente el uso de los equipos ni la calidad pedagógica. Además, es necesario saber si los locales conectados presentan también equipos funcionales y si los no conectados se encuentran en contextos de mayor desventaja o aislamiento territorial.

<a name='objetivo'></a>

**Objetivo**

El objetivo es identificar, mediante clustering, perfiles de brecha tecnológica en los locales educativos del Perú, considerando su acceso a internet, la disponibilidad y el estado de sus recursos tecnológicos y la cobertura móvil de su entorno, y localizar dónde se concentran en el territorio para orientar acciones priorizadas.

**Pregunta de análisis.** ¿Qué perfiles de brecha tecnológica se distinguen entre los locales educativos del Perú al combinar estas tres dimensiones, y cómo se distribuyen y concentran en el territorio?

**Usuarios del análisis.** MINEDU, Direcciones Regionales de Educación y Unidades de Gestión Educativa Local.

**Sección 1: Importar librerías y cargar fuentes**

```python
# Importar librerías
from dbfread import DBF
import pandas as pd
import numpy as np
import seaborn as sns

# Ajustar tamaño de tablas
pd.set_option("display.max_columns", 100)
pd.set_option("display.max_rows", 100)

# Cargar fuentes .dbf
servicios_internet_raw = pd.DataFrame(iter(DBF(r"datos/Loc_p3230_inter.dbf", encoding="latin1")))
recursos_tecnologicos_raw = pd.DataFrame(iter(DBF(r"datos/loc_p91_rectec.dbf", encoding="latin1")))
padron_raw  = pd.DataFrame(iter(DBF(r"datos/Padlocal.dbf", encoding="latin1")))
censo_local_raw = pd.DataFrame(iter(DBF(r"datos/loc_lineal_1.dbf", encoding="latin1")))

# Cargar fuentes .excel y concatenar
osiptel_1 = pd.read_excel(r"datos/osiptel_cobertura_2025IV_parte1.xlsx", sheet_name="Listado", dtype={"Ubigeo": str})
osiptel_2 = pd.read_excel(r"datos/osiptel_cobertura_2025IV_parte2.xlsx", sheet_name="Listado", dtype={"Ubigeo": str})
cobertura_movil_raw = pd.concat([osiptel_1, osiptel_2], ignore_index=True)


# Mostrar los 3 primeros registros de cada fuente
print("DATASET: servicios_internet\n")
display(servicios_internet_raw.head(3))

print("DATASET: recursos_tecnologicos\n")
display(recursos_tecnologicos_raw.head(3))

print("DATASET: padron\n")
display(padron_raw.head(3))

print("DATASET: cobertura_movil\n")
display(cobertura_movil_raw.head(3))

print("DATASET: censo_local\n")
display(censo_local_raw.head(3))
```

**Sección 2: Selección de columnas a trabajar**

Seleccionamos las columnas relevantes para responder a la pregunta de investigación y descartamos aquellas que no aportan información relacionada con nuestro tema de estudio.

```python
servicios_internet = servicios_internet_raw.drop(columns=["NROCED", "CUADRO", "ANEXO"])
recursos_tecnologicos  = recursos_tecnologicos_raw.drop(columns=["NROCED", "CUADRO","GESTION", "ANEXO",'P91_3_411',
       'P91_3_412', 'P91_3_413', 'P91_3_4141', 'P91_3_4142', 'P91_3_421',
       'P91_3_422', 'P91_3_423', 'P91_3_4241', 'P91_3_4242', 'P91_3_431',
       'P91_3_432', 'P91_3_433', 'P91_3_4341', 'P91_3_4342'])
padron  = padron_raw.drop(columns=['CLOC_L1',"TIP_ENVIO"])
cobertura_movil = cobertura_movil_raw.drop(columns= [
       'Departamento', 'Provincia', 'Distrito', 'CentroPoblado',
       'Bitel_3G_CG+CAR', 'Bitel_4G_CG+CAR', 'Bitel_5G_CG+CAR',
       'Claro_2G_CG+CAR', 'Claro_3G_CG+CAR', 'Claro_4G_CG+CAR',
       'Claro_5G_CG+CAR', 'Entel_2G_CG+CAR', 'Entel_3G_CG+CAR',
       'Entel_4G_CG+CAR', 'Entel_5G_CG+CAR', 'Integratel_2G_CG+CAR',
       'Integratel_3G_CG+CAR', 'Integratel_4G_CG+CAR', 'Integratel_5G_CG+CAR'])
censo_local = censo_local_raw[["CODLOCAL", "DPTO", "CEN_POB", "LATITUD", "LONGITUD", "P311_5", "P3230", "P3230_SI"]].copy()
```

**Decisiones:**

* `Servicios_internet:` Se eliminan **NROCED**, **CUADRO** y **ANEXO**, y se conservan los campos que describen las líneas y su registro.
* `Recursos_tecnologicos:` Se eliminan **NROCED**, **CUADRO**, **GESTION** y **ANEXO**, además de las 15 columnas del bloque **P91_3_411 a P91_3_4342**. Se trabaja con los totales, el estado, la antigüedad y el tipo de recurso, sin el desglose por ambiente.
* `Padron:` Se eliminan **CLOC_L1** y **TIP_ENVIO**. Más adelante se seleccionan seis campos para la integración: código del local, distrito, centro poblado MINEDU, departamento, provincia y nombre del distrito.
* `Cobertura_movil:` Se conservan las columnas de cobertura garantizada **CG** y se excluyen las **CG+CAR**. También se retiran los nombres geográficos, ya que la unión se realiza mediante códigos. No se calcula una correlación entre CG y CG+CAR en este cuaderno.
* `Censo_local:` Se conservan ocho campos: código del local, región, centro poblado, coordenadas, acceso a internet del centro poblado y del local, y cantidad de líneas declaradas.


| Dataset | Columnas seleccionadas o eliminadas | Motivo |
|---|---|---|
| Servicios_internet | Se eliminan NROCED, CUADRO y ANEXO | Se conservan los atributos de las líneas |
| Recursos_tecnologicos | Se eliminan NROCED, CUADRO, GESTION, ANEXO y las 15 columnas P91_3_4xx | Se usan los totales y condiciones del equipamiento, sin el desglose por ambiente |
| Padron | Se eliminan CLOC_L1 y TIP_ENVIO | Se conserva la identificación y ubicación del local |
| Cobertura_movil | Se eliminan nombres geográficos y columnas CG+CAR | Se trabaja con cobertura garantizada y códigos territoriales |
| Censo_local | Se seleccionan ocho campos | Identificación, ubicación y conectividad del local |

La concentración de ceros ayuda a interpretar los resultados del IQR. En `cobertura_5g_promedio`, el 66,43 % de las filas de la base tiene valor cero; Q1 es 0 y Q3 es aproximadamente 0,001772. Por tanto, el IQR no es cero en esta ejecución. Su límite superior es aproximadamente 0,00443 y señala 9 227 locales. Se conservan esos valores: superar el umbral indica una diferencia respecto de la distribución, no demuestra un error de cobertura.


```python
print(base_final["cobertura_5g_promedio"].describe())
print((base_final["cobertura_5g_promedio"] == 0).mean())
```

Se aplican dos reglas de detección:

- **Grupo 1:** cantidades de recursos y cobertura media 5G. Se buscan valores por encima de Q3 + 1,5 × IQR.
- **Grupo 2:** velocidades, ancho de banda y cobertura media 4G. Se revisan los límites inferior y superior del IQR.

Esta separación describe la regla aplicada, no implica que todas las variables del primer grupo tengan muchos ceros: la cantidad total de recursos no tiene ceros en la base actual. Las marcas sirven para revisar casos; el código no elimina automáticamente todos los registros señalados.


```python
def detectar_outliers_iqr(df, columna):
    q1 = df[columna].quantile(0.25)
    q3 = df[columna].quantile(0.75)
    iqr = q3 - q1
    lim_inf = q1 - 1.5 * iqr
    lim_sup = q3 + 1.5 * iqr
    outliers = df[(df[columna] < lim_inf) | (df[columna] > lim_sup)]
    return outliers, lim_inf, lim_sup

def detectar_outliers_superior(df, columna):
    q1 = df[columna].quantile(0.25)
    q3 = df[columna].quantile(0.75)
    iqr = q3 - q1
    lim_sup = q3 + 1.5 * iqr
    outliers = df[df[columna] > lim_sup]
    return outliers, lim_sup
```

```python
grupo_1 = ["cantidad_necesita_reparacion", "cobertura_5g_promedio", "numero_recursos_menos_3_años",
           "numero_recursos_mas_3_años", "cantidad_buen_estado", "cantidad_recur_operativo",
           "cantidad_total_recursos"]

grupo_2 = ["minimo_ancho_banda", "maximo_ancho_banda", "max_velocidad_bajada", "min_velocidad_bajada",
           "max_velocidad_subida", "min_velocidad_subida", "cobertura_4g_promedio"]

resumen_outliers = {}

for col in grupo_1:
    outliers, lim_sup = detectar_outliers_superior(base_final, col)
    resumen_outliers[col] = len(outliers)

for col in grupo_2:
    outliers, lim_inf, lim_sup = detectar_outliers_iqr(base_final, col)
    resumen_outliers[col] = len(outliers)

pd.Series(resumen_outliers).sort_values(ascending=False)
```

**Sección 10: Transformación de variables y construcción de la matriz analítica**

Después de la limpieza, integración de las fuentes y revisión de valores atípicos, se preparan las variables que posteriormente serán utilizadas para identificar perfiles de brecha tecnológica.

La técnica propuesta es **clustering**, ya que se busca agrupar locales educativos con características tecnológicas similares sin disponer de una variable objetivo previamente definida.

Para ello se realizan tres pasos:

1. Crear variables que permitan comparar locales de diferente tamaño.
2. Seleccionar las variables que realmente representan la brecha tecnológica.
3. Reescalar únicamente las variables cuantitativas que utilizan escalas diferentes.


**Transformación de variables**

Se crea una copia de `base_final` para conservar intacta la base integrada original.

Las variables que indican presencia de Internet se transforman de "Sí/No" a valores numéricos:

- Sí = 1
- No = 0

Los valores desconocidos permanecen como faltantes y no se interpretan como ausencia de Internet.

```python
base_modelo = base_final.copy()

# Sí = 1, No = 0
base_modelo["local_educativo_tiene_internet"] = (
    base_modelo["local_educativo_tiene_internet"]
    .map({"Sí": 1, "No": 0})
)

base_modelo["centro_poblado_tiene_internet"] = (
    base_modelo["centro_poblado_tiene_internet"]
    .map({"Sí": 1, "No": 0})
)
```

**Creación de variables derivadas**

Además de la cantidad total, se calculan tres proporciones sobre el inventario del local:

- `prop_operativos`: recursos operativos divididos entre el total.
- `prop_reparacion`: recursos que necesitan reparación divididos entre el total. En esta fuente forman parte de los operativos.
- `prop_antiguos`: recursos operativos con más de tres años divididos entre el total. No describe la antigüedad de los recursos no operativos.

Estas proporciones comparan la composición del inventario; no miden si la cantidad de equipos es suficiente para el número de estudiantes.


```python
total = base_modelo["cantidad_total_recursos"]

base_modelo["prop_operativos"] = (
    base_modelo["cantidad_recur_operativo"] / total
)

base_modelo["prop_reparacion"] = (
    base_modelo["cantidad_necesita_reparacion"] / total
)

base_modelo["prop_antiguos"] = (
    base_modelo["numero_recursos_mas_3_años"] / total
)
```

**Selección de variables para la matriz analítica**

La matriz analítica contiene únicamente las variables que deben intervenir en la formación de los perfiles tecnológicos.

Se consideran cinco dimensiones:

- disponibilidad de recursos;
- estado y antigüedad;
- tipos de recursos;
- conectividad del local;
- cobertura móvil territorial.

## **3. Metodología**

El clustering agrupa locales con características parecidas para describir sus perfiles. Antes de agruparlos, se aplica `log1p` al total de recursos y al número de líneas de internet, se estandarizan las once variables cuantitativas y se mantienen las seis binarias en 0/1. El PCA resume la información de las 17 variables en componentes que sirven de entrada a los métodos de agrupamiento.

El análisis territorial responde dos preguntas. El mapa departamental muestra qué perfil predomina en cada departamento. El DBSCAN espacial identifica dónde se concentran los locales del perfil G3 a partir de sus coordenadas.

### **3.1. Transformación de variables y análisis de componentes principales**

```python
import numpy as np
from sklearn.preprocessing import StandardScaler

# 1. Aplicar logaritmo solo a las cantidades
columnas_log = ["cantidad_total_recursos", "cantidad_lineas_internet"]

matriz_log = matriz_analitica.dropna().copy()
for col in columnas_log:
    matriz_log[col] = np.log1p(matriz_log[col])

# 2. Estandarizar todo menos las binarias
matriz_pca = matriz_log.copy()
scaler = StandardScaler()
matriz_pca[columnas_reescalar] = scaler.fit_transform(matriz_log[columnas_reescalar])

print("Dimensiones para PCA:", matriz_pca.shape)
```

```python
from sklearn.decomposition import PCA

pca = PCA()
Z = pca.fit_transform(matriz_pca)
P = pca.components_.T

variables = matriz_pca.columns.tolist()
nombres_z = [f"Z{i+1}" for i in range(len(variables))]

tabla_varianza = pd.DataFrame({
    "Componente": nombres_z,
    "Porcentaje": 100 * pca.explained_variance_ratio_,
    "Acumulada": 100 * pca.explained_variance_ratio_.cumsum()
})
display(tabla_varianza.round(2))

n_80 = (tabla_varianza["Acumulada"] >= 80).idxmax() + 1
print("Componentes para llegar al 80 %:", n_80)
```

```python
fig, ax = plt.subplots(figsize=(10, 4))
ax.bar(tabla_varianza["Componente"], tabla_varianza["Porcentaje"], label="Por componente")
ax.plot(tabla_varianza["Componente"], tabla_varianza["Acumulada"], color="black", marker="o", label="Acumulada")
ax.axhline(80, color="red", linestyle="--", label="80 %")
ax.set(title="Varianza explicada por PCA (con log)", ylabel="Varianza explicada (%)", ylim=(0, 105))
ax.legend()
plt.show()
```

**Lectura del PCA.** Los cinco primeros componentes explican el 84,4 % de la varianza. Son la entrada de K-means, Ward y DBSCAN en la comparación de perfiles.

### **3.2. Pesos e interpretación de los componentes**

## **4. Diseño experimental**

El clustering se aplica sobre `Z_df`, que contiene los 5 componentes del PCA (Z1 a Z5, 84.4 % de la varianza). Así se trabaja con menos dimensiones y sin variables correlacionadas.

Se prueban tres técnicas y al final se elige una:

1. **Prueba 1: K-means**, con k de 2 a 8.
2. **Prueba 2: Jerárquico Ward.**
3. **Prueba 3: DBSCAN.**
4. **Decisión:** técnica elegida y perfiles de los grupos (resultados preliminares).

Los atípicos se conservan porque son parte de la realidad de los colegios peruanos. El logaritmo que se aplicó antes del PCA suaviza su efecto en las dos cantidades (recursos y líneas).

| Técnica | Configuraciones evaluadas | Criterio de lectura |
|---|---|---|
| K-means | De 2 a 8 grupos | Inercia, silueta y Davies–Bouldin |
| Ward | Dendrograma con 5 000 locales y corte en altura 60 | Silueta y Davies–Bouldin para el corte de cinco grupos |
| DBSCAN sobre PCA | `eps` de 0,3 a 0,7 con `min_samples = 10` | Número de grupos, proporción de ruido y tamaño del grupo mayor |

En la silueta se buscan valores más altos y en Davies–Bouldin, valores más bajos.

### **4.1. Prueba de K-means**

Se prueban de 2 a 8 grupos y se evalúan con tres criterios, como en el curso: el codo de la inercia, la silueta (más alta, mejor) y el Davies-Bouldin (más bajo, mejor). La silueta se calcula con una muestra de 10 000 locales por el tamaño de la base.

```python
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score, davies_bouldin_score

filas = []
for k in range(2, 9):
    km = KMeans(n_clusters=k, n_init=10, random_state=42).fit(Z_df)
    filas.append({"k": k, "inercia": km.inertia_,
                  "silueta": silhouette_score(Z_df, km.labels_, sample_size=10000, random_state=42),
                  "davies_bouldin": davies_bouldin_score(Z_df, km.labels_)})
evaluacion = pd.DataFrame(filas)
display(evaluacion.round(3))

fig, axes = plt.subplots(1, 3, figsize=(14, 3.5))
for ax, col in zip(axes, ["inercia", "silueta", "davies_bouldin"]):
    sns.lineplot(data=evaluacion, x="k", y=col, marker="o", ax=ax); ax.set_title(col)
plt.tight_layout(); plt.show()
```

**Resultado Prueba 1:** k=2 obtiene la mejor silueta (0.505) y el menor Davies-Bouldin (0.793), pero solo divide la base en dos grupos muy generales. Entre k=3 y k=8, k=5 es el mejor en ambos criterios (silueta 0.331, Davies-Bouldin 1.073), k=6 queda casi igual (0.330 y 1.081) y la inercia baja de forma gradual, sin un codo claro. Se toma **k=5** como candidato de K-means.

### **4.2. Prueba de agrupamiento jerárquico Ward**

Se arma el dendrograma con el método Ward y se corta donde hay un salto grande de altura. Se usa una muestra de 5 000 locales porque con los 42 215 la memoria no alcanza (Ward compara todos los pares).

```python
from scipy.cluster.hierarchy import linkage, dendrogram

muestra = Z_df.sample(5000, random_state=42)
Z_ward = linkage(muestra, method="ward")

fig, ax = plt.subplots(figsize=(12, 4))
dendrogram(Z_ward, truncate_mode="lastp", p=30, ax=ax)
ax.set_title("Dendrograma Ward (muestra 5 000)")
plt.show()
```

**Lectura del dendrograma:** el salto más grande (altura ≈ 220) separa 2 grupos. Para comparar con K-means se corta en altura 60, lo que da 5 grupos (entre las uniones de ≈71 y ≈56).

```python
from scipy.cluster.hierarchy import fcluster

corte = 60
grupos_ward = fcluster(Z_ward, t=corte, criterion="distance")

print("Tamaño de cada grupo:")
print(pd.Series(grupos_ward).value_counts().sort_index())
print("Silueta:", round(silhouette_score(muestra, grupos_ward), 3))
print("Davies-Bouldin:", round(davies_bouldin_score(muestra, grupos_ward), 3))
```

**Resultado Prueba 2:** con el corte en 60, Ward forma 5 grupos (entre 241 y 1 575 locales de la muestra). Obtiene silueta 0.289 y Davies-Bouldin 1.139, ambos algo peores que K-means con k=5 (0.331 y 1.073).

### **4.3. Prueba de DBSCAN sobre los componentes**

Primero se dibuja la curva de k-distancias (10 vecinos) para ver hasta qué `eps` tiene sentido probar. El codo marca el `eps` máximo razonable.

```python
from sklearn.neighbors import NearestNeighbors

vecinos = NearestNeighbors(n_neighbors=10).fit(Z_df)
distancias, _ = vecinos.kneighbors(Z_df)
k_dist = np.sort(distancias[:, -1])

plt.figure(figsize=(8, 3.5))
plt.plot(k_dist)
plt.title("Curva de k-distancias (10 vecinos)")
plt.ylabel("distancia al 10.º vecino")
plt.show()
```

**Lectura de k-distancias:** la curva es suave hasta ≈0.6 y forma el codo entre 0.7 y 0.8. Se prueban valores de `eps` de 0.3 a 0.7 con `min_samples=10`.

```python
from sklearn.cluster import DBSCAN

filas_db = []
for eps in [0.3, 0.4, 0.5, 0.6, 0.7]:
    etiquetas = DBSCAN(eps=eps, min_samples=10).fit_predict(Z_df)
    grupos = len(set(etiquetas)) - (1 if -1 in etiquetas else 0)
    ruido = (etiquetas == -1).mean()
    sin_ruido = etiquetas[etiquetas != -1]
    mas_grande = pd.Series(sin_ruido).value_counts(normalize=True).max() if grupos > 0 else 0
    filas_db.append({"eps": eps, "grupos": grupos, "ruido": round(ruido, 3),
                     "grupo_mas_grande": round(mas_grande, 3)})

display(pd.DataFrame(filas_db))
```

Con `eps` cercano al codo, un solo grupo absorbe casi todo (eps=0.7: 99.6 %). Los datos no tienen regiones densas separadas, por lo que DBSCAN no produce perfiles útiles y se descarta.

### **5.5. Focos espaciales del perfil G3**

Se aplica DBSCAN sobre las coordenadas de los locales del perfil G3 (inventario pequeño y desconectado) para ver si forman focos concentrados en el territorio. Se usa la distancia haversine, que mide distancias reales sobre la Tierra, y primero se revisa la curva de k-distancia en kilómetros para elegir el radio.

#### **5.5.1. Comparación de radios en DBSCAN espacial**

```python
from sklearn.neighbors import NearestNeighbors
import numpy as np
import matplotlib.pyplot as plt

g3 = gdf_locales[gdf_locales["clase"] == "G3 Inventario pequeño y desconectado"].copy()
print("Locales G3:", len(g3))

coords = np.radians(np.column_stack([g3.geometry.y, g3.geometry.x]))

nn = NearestNeighbors(n_neighbors=10, metric="haversine", algorithm="ball_tree")
nn.fit(coords)
dist, _ = nn.kneighbors(coords)

k_dist_km = np.sort(dist[:, -1]) * 6371

plt.plot(k_dist_km)
plt.ylim(0, 50)
plt.xlabel("Locales ordenados")
plt.ylabel("Distancia al 10.º vecino (km)")
plt.title("k-distancia G3")
plt.show()
```

**Lectura:** G3 tiene 13 241 locales. La distancia al 10.º vecino crece suave hasta unos 10–12 km y se dispara después de ~15 km, donde quedan los locales aislados. Se prueban radios entre 5 y 20 km.

```python
from sklearn.cluster import DBSCAN

radios = [5, 7.5, 10, 12.5, 15, 20]
filas = []

for km in radios:
    db = DBSCAN(eps=km / 6371, min_samples=10, metric="haversine", algorithm="ball_tree")
    etiquetas = db.fit_predict(coords)
    n_grupos = len(set(etiquetas)) - (1 if -1 in etiquetas else 0)
    ruido = (etiquetas == -1).mean()
    tam = pd.Series(etiquetas[etiquetas != -1]).value_counts()
    mayor = tam.max() / len(etiquetas) if len(tam) > 0 else 0
    filas.append([km, n_grupos, round(ruido, 3), round(mayor, 3)])

tabla_radios = pd.DataFrame(filas, columns=["radio_km", "focos", "ruido", "foco_mas_grande"])
display(tabla_radios)
```

**Lectura:** Con 5 y 7.5 km el ruido es muy alto (63 % y 34 %). Desde 15 km aparece un foco que concentra la mayor parte del perfil (68 % a 15 km y 87 % a 20 km), lo que ya no describe focos locales. Se elige **10 km**: 124 focos, 19.3 % de ruido y un foco mayor con el 30.4 % de los locales G3.

Esto sugiere que, para políticas de conectividad, la sierra permite intervenciones por zonas y la selva requiere atención local por local.

La sección 6 extiende esta lectura a los cinco perfiles, la valida con el área y la gestión, y propone una acción prioritaria por región, provincia y UGEL. Los focos todavía no se han contrastado con la densidad del conjunto de locales.

```python
import json
import plotly.graph_objects as go
import plotly.colors as pc

# 1. Área, gestión y UGEL desde el padrón (1 = urbano, 2 = rural; A = pública, B = privada)
extra = padron[["cod_local", "area_codigo", "gestion_codigo", "cod_ugel", "nombre_ugel"]].copy()
extra["area"] = extra["area_codigo"].map({"1": "Urbano", "2": "Rural"})
extra["gestion"] = extra["gestion_codigo"].map({"A": "Pública", "B": "Privada"})
locales_h4 = gdf_locales.merge(extra.drop(columns=["area_codigo", "gestion_codigo"]),
                               on="cod_local", how="left", validate="one_to_one")
locales_h4["perfil"] = locales_h4["clase"].str[:2]
locales_h4["prov"] = locales_h4["index_right"].astype("Int64").astype("string")   # provincia (sección 5.4)
locales_h4.loc[locales_h4["prov"].isna(), "dep_geo"] = pd.NA   # 9 locales fuera de todo polígono (sección 5.4)

perfiles = sorted(locales_h4["perfil"].unique())
nombres = {p: nombres_grupos[int(p[1])] for p in perfiles}
colores = colores_perfil   # G0 dorado, G1 azul, G2 naranja, G3 verde, G4 morado (sección 5.4.2)
nacional = locales_h4["perfil"].value_counts(normalize=True)

# 2. Regiones y provincias en GeoJSON (contornos simplificados a ~2.5 km)
regiones = gdf_dep[["dep_geo", "ADM1_ES", "geometry"]].copy()
provincias = gdf_provincias[["ADM1_ES", "ADM2_ES", "geometry"]].copy()
provincias["prov"] = provincias.index.astype(str)

def a_geojson(gdf, clave):
    g = gdf[[clave, "geometry"]].copy()
    g["geometry"] = g.simplify(0.025).set_precision(0.01)
    return json.loads(g.to_json())

ubicaciones = {
    "Región": dict(clave="dep_geo", geojson=a_geojson(regiones, "dep_geo"),
                   nombre=regiones.set_index("dep_geo")["ADM1_ES"]),
    "Provincia": dict(clave="prov", geojson=a_geojson(provincias, "prov"),
                      nombre=provincias.set_index("prov")["ADM2_ES"] + " (" + provincias.set_index("prov")["ADM1_ES"] + ")"),
}
punto_region = regiones.set_index("dep_geo").representative_point()

def geojson_de(ubicacion, ids):
    # Solo los polígonos que pinta la capa, para que la figura no pese de más
    u = ubicaciones[ubicacion]
    ids = set(map(str, ids))
    return {"type": "FeatureCollection",
            "features": [f for f in u["geojson"]["features"] if str(f["properties"][u["clave"]]) in ids]}   # para escribir valores sobre cada región

# 3. Capas y diseño comunes a los tres mapas
def capa_base():
    return go.Choropleth(geojson=ubicaciones["Región"]["geojson"], featureidkey="properties.dep_geo",
                         locations=regiones["dep_geo"], z=[0] * len(regiones),
                         colorscale=[[0, "#eeeeee"], [1, "#eeeeee"]], showscale=False, hoverinfo="skip",
                         marker_line_color="white", marker_line_width=0.5)

def capa_contorno():
    return go.Choropleth(geojson=ubicaciones["Región"]["geojson"], featureidkey="properties.dep_geo",
                         locations=regiones["dep_geo"], z=[0] * len(regiones),
                         colorscale=[[0, "rgba(0,0,0,0)"], [1, "rgba(0,0,0,0)"]], showscale=False, hoverinfo="skip",
                         marker_line_color="#555555", marker_line_width=0.8)

def menu_ubicacion(trazas_region, trazas_provincia, x=0.01):
    # Región o provincia: cambia la opacidad y el hover de esas capas, sin tocar lo que eligen los otros menús
    def args(region):
        activas = [region] * len(trazas_region) + [not region] * len(trazas_provincia)
        return [{"marker.opacity": [0.8 if a else 0 for a in activas], "opacity": [1 if a else 0 for a in activas],
                 "hoverinfo": ["text" if a else "skip" for a in activas], "showscale": activas},
                trazas_region + trazas_provincia]
    return dict(buttons=[dict(label="Región", method="restyle", args=args(True)),
                         dict(label="Provincia", method="restyle", args=args(False))],
                direction="down", x=x, y=1.0, xanchor="left", yanchor="top")

def escala_discreta(lista_colores):
    # Una franja de color por categoría (valores 0, 1, 2, ... con zmin = -0.5 y zmax = n - 0.5)
    n = len(lista_colores)
    return [[f, c] for i, c in enumerate(lista_colores) for f in (i / n, (i + 1) / n)]

def configurar_mapa(fig, titulo, menus):
    fig.update_geos(
        visible=False, lonaxis_range=[-82, -68], lataxis_range=[-18.6, 0.2],
        showland=True, landcolor="#f4f4f4", showocean=True, oceancolor="#e6f0f7",
        showcountries=True, countrycolor="#bbbbbb"
    )
    fig.update_layout(
        height=720, margin=dict(l=0, r=0, t=110, b=0),
        title={"text": titulo, "x": 0.01, "y": 0.985},
        template="plotly_white", hoverlabel=dict(bgcolor="white", font_size=13),
        updatemenus=[menu for _, menu in menus],
        annotations=[dict(text=f"<b>{nombre}</b>", x=menu["x"], y=1.07, xref="paper", yref="paper",
                          showarrow=False, xanchor="left") for nombre, menu in menus],
        showlegend=False
    )

print("Locales:", len(locales_h4), "| faltantes en área, gestión y UGEL:", locales_h4[["area", "gestion", "cod_ugel"]].isna().sum().sum())
cuenta_prov = pd.crosstab(locales_h4["prov"], locales_h4["perfil"])
indice_prov = cuenta_prov.div(cuenta_prov.sum(axis=1), axis=0) / nacional
print("Provincias donde predomina cada perfil:", cuenta_prov.idxmax(axis=1).value_counts().reindex(perfiles, fill_value=0).to_dict())
print("Provincias donde está sobrerrepresentado:", (indice_prov >= 1).sum().to_dict())
```

### **6.1. Focos y locales aislados de cada perfil**

Un local en **foco** forma parte de una zona densa de locales del mismo perfil (al menos 10 en un radio de 10 km) y se puede atender por zonas; un local **aislado** (ruido de DBSCAN) se atiende local por local.

**Radio común.** Se usa el mismo radio para los cinco perfiles: 10 km, elegido con la curva de k-distancia de G3 en la sección 5.5 (con menos radio casi todo G3 queda aislado; con más, aparece un foco que reúne la mayor parte del perfil). Con un mismo radio, un foco exige la misma densidad en todos los perfiles y el % de aislados se puede comparar entre ellos. Para comprobar que la conclusión no depende de los 10 km, se repite DBSCAN con radios de 5 a 20 km (`min_samples` = 10).

```python
def coordenadas(sub):
    return np.radians(np.column_stack([sub.geometry.y, sub.geometry.x]))

def dbscan_espacial(sub, km, min_samples=10):
    db = DBSCAN(eps=km / 6371, min_samples=min_samples, metric="haversine", algorithm="ball_tree")
    return db.fit_predict(coordenadas(sub))

radio = 10   # radio común, elegido con la curva de G3 (sección 5.5)

filas = []
fig, ax = plt.subplots(figsize=(8, 4.5))
for p in perfiles:
    sub = locales_h4[locales_h4["perfil"] == p]
    dist, _ = NearestNeighbors(n_neighbors=10, metric="haversine", algorithm="ball_tree").fit(coordenadas(sub)).kneighbors(coordenadas(sub))
    k_km = np.sort(dist[:, -1]) * 6371
    ax.plot(np.linspace(0, 100, len(k_km)), k_km, color=colores[p], label=nombres[p])
    for km in [5, 7.5, 10, 12.5, 15, 20]:
        et = dbscan_espacial(sub, km)
        tam = pd.Series(et[et != -1]).value_counts()
        filas.append([p, km, len(tam), 100 * (et == -1).mean(), 100 * tam.max() / len(et)])

ax.axhline(radio, color="grey", linestyle="--", linewidth=1)
ax.set_ylim(0, 40)
ax.set_xlabel("Locales del perfil, ordenados (%)")
ax.set_ylabel("Distancia al 10.º vecino (km)")
ax.set_title("k-distancia por perfil (línea punteada: radio común de 10 km)")
ax.legend(fontsize=7)
plt.savefig("salidas/hito4_distancias_por_perfil.png", dpi=150, bbox_inches="tight")
plt.show()

radios = pd.DataFrame(filas, columns=["perfil", "radio_km", "focos", "% aislados", "% foco mayor"]).round(1)
radios.to_csv("salidas/hito4_resumen_focos.csv", index=False, encoding="utf-8-sig")
display(radios.pivot(index="radio_km", columns="perfil", values=["% aislados", "% foco mayor"]))
```

**Lectura:** La curva de k-distancia muestra la diferencia de densidad: la mitad de los locales G3 tiene a su 10.º vecino del mismo perfil a menos de 7.4 km, la de G1 a 11.3 km y la de G4 a 15.9 km, mientras que G0 y G2 están mucho más juntos (0.5 y 2.0 km). La línea de 10 km deja por debajo al 69 % de los locales G3, al 40 % de G1 y solo al 16 % de G4. En la tabla, con cualquier radio entre 5 y 20 km el orden se mantiene: G2 y G3 tienen la menor proporción de locales aislados, G1 queda en medio y G4 es el más disperso. A 10 km quedan aislados el 19.3 % de G3, el 42.5 % de G1 y el 69.2 % de G4. Para que G4 tuviera tan pocos aislados como G3 habría que llevar el radio a 20 km (21.4 %), es decir, aceptar como foco zonas cuatro veces menos densas. Con 10 km ningún foco de G1, G2 ni G4 reúne más del 9 % del perfil; el mayor de G3 reúne el 30.4 %. G0 no forma focos de brecha: con cualquier radio, un solo foco (Lima y Callao) reúne el 83.8 % del perfil, porque es urbano y continuo.

```python
# 1. Focos y locales aislados de cada perfil con el radio común
locales_h4["foco"] = -1
for p in perfiles:
    m = locales_h4["perfil"] == p
    locales_h4.loc[m, "foco"] = dbscan_espacial(locales_h4[m], radio)
locales_h4["en_foco"] = locales_h4["foco"] != -1

g3_h4 = locales_h4[locales_h4["perfil"] == "G3"]
print("Control G3:", g3_h4["foco"].max() + 1, "focos y", (~g3_h4["en_foco"]).sum(), "aislados (sección 5.5: 124 y 2 556)")

# 2. Locales en focos y aislados por región, para los perfiles de brecha
brecha = ["G1", "G3", "G4"]
locales_h4["region"] = locales_h4["dep_geo"].map(ubicaciones["Región"]["nombre"])
locales_h4["lon"], locales_h4["lat"] = locales_h4.geometry.x, locales_h4.geometry.y
por_region = (locales_h4[locales_h4["perfil"].isin(brecha)]
              .groupby(["perfil", "region"])["en_foco"].agg(locales="size", en_focos="sum"))
por_region["aislados"] = por_region["locales"] - por_region["en_focos"]
por_region["% aislados"] = (100 * por_region["aislados"] / por_region["locales"]).round(0)
display(por_region["% aislados"].unstack("perfil").sort_values("G3"))

# 3. Zona de cada foco: unión de un área de 5 km (la mitad del radio) alrededor de sus locales
def regiones_principales(s):
    return ", ".join(f"{r} ({n})" for r, n in s.value_counts().head(3).items())

en_focos = locales_h4[locales_h4["perfil"].isin(brecha) & locales_h4["en_foco"]].to_crs(32718)   # UTM 18S, en metros
zonas = gpd.GeoDataFrame(en_focos[["perfil", "foco", "region"]].assign(locales=1),
                         geometry=en_focos.buffer(radio * 500, quad_segs=4), crs=en_focos.crs)
zonas = zonas.dissolve(["perfil", "foco"], aggfunc={"locales": "sum", "region": regiones_principales})
zonas["geometry"] = zonas.simplify(500)
zonas = zonas.to_crs(4326).reset_index()
zonas["id"] = zonas["perfil"] + "-" + zonas["foco"].astype(str)
print("Focos por perfil:", zonas["perfil"].value_counts().reindex(brecha).to_dict())

# 4. Detalle por provincia y ficha de cada local con su ubicación
por_provincia = (locales_h4[locales_h4["perfil"].isin(brecha)]
                 .groupby(["perfil", "prov"])["en_foco"].agg(locales="size", en_focos="sum"))
por_provincia["aislados"] = por_provincia["locales"] - por_provincia["en_focos"]

locales_h4["direccion"] = locales_h4["cod_local"].map(padron.set_index("cod_local")["direccion"])
ficha_locales = locales_h4[locales_h4["perfil"].isin(brecha)].assign(
    foco=lambda d: d["foco"].where(d["en_foco"]).astype("Int64"),
    atencion=lambda d: np.where(d["en_foco"], "Por zonas", "Local por local"),
    lat=lambda d: d["lat"].round(5), lon=lambda d: d["lon"].round(5))
ficha_locales = ficha_locales[["cod_local", "perfil", "atencion", "foco", "departamento", "provincia", "distrito",
                               "centro_poblado", "direccion", "nombre_ugel", "lat", "lon"]]
ficha_locales.sort_values(["perfil", "departamento", "provincia", "distrito"]).to_csv(
    "salidas/hito4_locales_focos.csv", index=False, encoding="utf-8-sig")
print("Ficha de locales (G1, G3 y G4):", len(ficha_locales), "→ salidas/hito4_locales_focos.csv")
```

### **6.2. Validación con área y gestión**

El área y la gestión no entraron al clustering. Si la frecuencia de los perfiles cambia mucho entre locales rurales y urbanos, o entre públicos y privados, los perfiles reflejan diferencias reales y no solo un resultado del método. Se revisa en dos niveles: por local, con barras de la frecuencia de cada perfil dentro de cada categoría (V de Cramér), y por provincia, con cuatro mapas en la misma escala (correlación de Spearman). La gestión se compara solo entre locales urbanos, porque casi no hay privados rurales (100 de 24 192).

```python
from scipy.stats import chi2_contingency

# 1. Asociación entre perfil, área y gestión (V de Cramér)
def v_cramer(a, b):
    tabla = pd.crosstab(a, b)
    chi2, p, _, _ = chi2_contingency(tabla)
    return np.sqrt(chi2 / (tabla.values.sum() * (min(tabla.shape) - 1))), p

urbanos = locales_h4[locales_h4["area"] == "Urbano"]   # la gestión se compara solo entre urbanos
v_area, p_area = v_cramer(locales_h4["perfil"], locales_h4["area"])
v_gestion, p_gestion = v_cramer(urbanos["perfil"], urbanos["gestion"])

# 2. Indicadores por provincia: % rural, % en G1/G3/G4, % privada y % en G0/G2 (los dos últimos entre urbanos)
def indicadores(ubicacion):
    clave = ubicaciones[ubicacion]["clave"]
    cuenta = pd.crosstab(locales_h4[clave], locales_h4["perfil"])
    urb = urbanos.groupby(clave)
    t = pd.DataFrame({
        "locales": cuenta.sum(axis=1),
        "pct_rural": 100 * locales_h4.groupby(clave)["area"].apply(lambda s: (s == "Rural").mean()),
        "pct_atender": 100 * cuenta[["G1", "G3", "G4"]].sum(axis=1) / cuenta.sum(axis=1),
        "urbanos": urb.size(),
        "pct_privada": 100 * urb["gestion"].apply(lambda s: (s == "Privada").mean()),
        "pct_G0_G2": 100 * urb["perfil"].apply(lambda s: s.isin(["G0", "G2"]).mean()),
    })
    t.loc[t["urbanos"].fillna(0) < 10, ["pct_privada", "pct_G0_G2"]] = np.nan   # menos de 10 urbanos: sin dato
    t["nombre"] = ubicaciones[ubicacion]["nombre"].reindex(t.index)
    return t

val = {"Provincia": indicadores("Provincia")}
r_area = val["Provincia"][["pct_rural", "pct_atender"]].corr("spearman").iloc[0, 1]
r_gestion = val["Provincia"][["pct_privada", "pct_G0_G2"]].dropna().corr("spearman").iloc[0, 1]
print(f"Perfil × área: V de Cramér = {v_area:.2f} | provincias, % rural vs % en G1, G3 y G4: Spearman = {r_area:.2f}")
print(f"Perfil × gestión (urbanos): V de Cramér = {v_gestion:.2f} | provincias, % privada vs % en G0 y G2: Spearman = {r_gestion:.2f}")
print("p < 0.001 en las dos pruebas:", max(p_area, p_gestion) < 0.001)

# 3. Frecuencia de los cinco perfiles dentro de cada categoría del padrón
fig, ejes = plt.subplots(1, 2, figsize=(12, 4.6), sharey=True, constrained_layout=True)
comparaciones = [
    (locales_h4, "area", ["Rural", "Urbano"], "Área del local"),
    (urbanos, "gestion", ["Pública", "Privada"], "Gestión de los locales urbanos"),
]
for ax, (datos, campo, categorias, titulo) in zip(ejes, comparaciones):
    # Porcentaje de cada perfil dentro de la categoría (cada categoría suma 100 %)
    cuenta = pd.crosstab(datos["perfil"], datos[campo]).reindex(index=perfiles, columns=categorias, fill_value=0)
    pct = cuenta.div(cuenta.sum()).mul(100).rename(index=nombres)
    pct.columns = [f"{c} (n = {cuenta[c].sum():,})".replace(",", " ") for c in categorias]
    pct.plot.barh(ax=ax, color=["#315875", "#aec9d5"], width=0.78)
    # Valor al final de cada barra, con coma decimal
    for barras in ax.containers:
        etiquetas = ["<0,1" if 0 < x < 0.1 else f"{x:.1f}".replace(".", ",") for x in barras.datavalues]
        ax.bar_label(barras, labels=etiquetas, padding=3, fontsize=9)
    ax.set(xlim=(0, 100), xlabel="Porcentaje dentro de cada categoría", ylabel="")
    ax.set_title(titulo, fontweight="bold")
    ax.legend(loc="lower right", frameon=False, fontsize=9)
    ax.spines[["top", "right"]].set_visible(False)
ejes[0].invert_yaxis()   # G0 arriba, G4 abajo
plt.show()
```

**Lectura:**
- **Área (V de Cramér = 0.67, asociación fuerte):** G3 reúne el 45.4 % de los locales rurales frente al 12.5 % de los urbanos, y G1 y G4 siguen el mismo patrón (23.8 % y 14.9 % en los rurales; 5.4 % y 2.7 % en los urbanos). G0 casi no existe en el campo (<0.1 %) y es el 39.5 % de los urbanos; G2 sube de 15.9 % a 39.9 %. En total, el 84.1 % de los locales rurales está en G1, G3 o G4, frente al 20.6 % de los urbanos.

### **6.3. Acción prioritaria**

**La idea en una frase.** Cada local ya tiene un perfil (G0 a G4). Esta sección responde qué necesidad pesa más en cada territorio y cuántos locales hay que atender, y lo muestra en un mapa, un ranking (Pareto) y una tabla por UGEL.

**Paso a paso**

1. **De perfil a necesidad.** Solo tres perfiles generan una acción; G0 y G2 no generan acción.

   | Perfil | Necesidad |
   |---|---|
   | G3 | Conectividad |
   | G1 | Mantenimiento |
   | G4 | Reparación |

2. **Contar por territorio.** `pd.crosstab` arma una tabla con una fila por región, provincia o UGEL y una columna por perfil; cada celda es el número de locales. De ahí salen los conteos de conectividad, mantenimiento y reparación, el total a atender (G1 + G3 + G4) y el % a atender.

3. **Acción principal de cada territorio.** La función `accion_principal` aplica una regla simple:
   - Mira solo G1, G3 y G4.
   - Se queda con las necesidades que son más frecuentes en ese territorio que en el promedio del país (índice ≥ 1).
   - Entre ellas, gana la que tiene más locales.
   - Si ninguna es más frecuente que en el país, el territorio queda como "Sin necesidad dominante".

   *Ejemplo:* en el país el 31.4 % de los locales está en G3; en Loreto es el 39.6 %, así que conectividad está sobrerrepresentada. En Loreto también lo están mantenimiento (23.6 % frente a 15.9 %) y reparación (14.9 % frente a 9.7 %), pero gana conectividad porque tiene más locales (318, frente a 190 y 120).

**Los cuatro bloques**
- **Bloque A, mapa** (`salidas/hito4_mapa_accion.html`). Tres menús: **Nivel** (región o provincia), **Qué mostrar** (acción principal, una necesidad o el % a atender) y **Enfocar departamento** (acerca el mapa, atenúa el resto y marca el borde del elegido). Al elegir una necesidad, solo se colorea donde esa es la acción principal: más oscuro significa más locales y el gris es otra acción. Al pasar el cursor se ve la acción principal, los locales de cada necesidad y cuántos hay a atender. Al final genera la ficha UGEL con la misma regla (`salidas/hito4_ficha_ugel.csv`).
- **Bloque B, Pareto (¿qué atiendo primero?).** Barras con los 15 territorios con más locales a atender, de mayor a menor, y una línea con el % acumulado del total del país. Su menú cambia entre región o provincia y entre el total o cada necesidad; al pasar el cursor se ve también la acción principal y el total de locales de cada territorio.
- **Bloque C, tabla UGEL (¿qué me toca a mí?).** Una fila por UGEL con su acción principal, sus conteos y el %, con filtro por departamento.
- **Bloque D, panel final** (`salidas/hito4_panel_accion.html`). Junta una guía de lectura, el mapa, el Pareto y la tabla en una sola página que se abre en cualquier navegador, sin servidor ni internet.

**Quién lo usa y cómo**
- **MINEDU:** mapa por región → acción principal → % a atender → Pareto.
- **DRE:** mapa por provincia → enfocar su departamento → mirar cada necesidad.
- **UGEL:** tabla → filtrar su departamento → ubicar su fila.

**Dos aclaraciones**
- **Esta sección no usa DBSCAN.** Solo cuenta locales por territorio administrativo. DBSCAN (sección 6.1) agrupa locales por cercanía geográfica y responde otra pregunta.
- **A nivel región, un solo color resume muchas provincias.** Loreto sale "Conectividad" aunque una de sus provincias, Datem del Marañón, tiene como acción principal la reparación. Para ese detalle se cambia el nivel a provincia.

**Orden de ejecución:** bloque A, luego B, C y D. Si se cambia algo en A, hay que volver a correr B, C y D, porque usan `pri` y `ficha`, que se generan en A.

#### **Bloque A. Mapa 4: acción prioritaria y ficha UGEL**

```python
# =====================================================================
# 6.3. ACCIÓN PRIORITARIA - MAPA + FICHA UGEL
# =====================================================================
import numpy as np

SIN = "Sin necesidad dominante"
acciones = {"G3": "Conectividad", "G1": "Mantenimiento", "G4": "Reparación"}
lista_acciones = ["Conectividad", "Mantenimiento", "Reparación", SIN]
escala_accion = escala_discreta([colores["G3"], colores["G1"], colores["G4"], "#9e9e9e"])   # mismos colores que la 6.1
GRIS = "#e0e0e0"

# --- 1. Lógica de acción -------------------------------------------
def accion_principal(cuenta):
    """Perfil de acción con índice >= 1 (más frecuente que en el país) y más locales."""
    indice = cuenta.div(cuenta.sum(axis=1), axis=0) / nacional
    candidatos = cuenta[list(acciones)].where(indice[list(acciones)] >= 1)
    hay = candidatos.notna().any(axis=1)
    accion = pd.Series(SIN, index=cuenta.index)
    accion[hay] = candidatos[hay].idxmax(axis=1).map(acciones)
    return accion

def prioridad(ubicacion):
    clave = ubicaciones[ubicacion]["clave"]
    cuenta = pd.crosstab(locales_h4[clave], locales_h4["perfil"])
    t = pd.DataFrame({"locales": cuenta.sum(axis=1)})
    for perfil, accion in acciones.items():
        t[f"n_{accion}"] = cuenta[perfil]
    t["a_atender"] = cuenta[list(acciones)].sum(axis=1)
    t["pct_a_atender"] = (100 * t["a_atender"] / t["locales"]).round(1)
    t["accion"] = accion_principal(cuenta)
    t["codigo"] = t["accion"].map({a: i for i, a in enumerate(lista_acciones)})
    t["nombre"] = ubicaciones[ubicacion]["nombre"].reindex(t.index)
    return t

def texto_hover(t):
    extra = t["accion"].eq(SIN).map({True: "<br><i>Ninguna necesidad supera al promedio del país</i>", False: ""})
    return ("<b>" + t["nombre"] + "</b>"
            + "<br>Acción principal: <b>" + t["accion"] + "</b>" + extra
            + "<br>Conectividad: " + t["n_Conectividad"].astype(str)
            + "<br>Mantenimiento: " + t["n_Mantenimiento"].astype(str)
            + "<br>Reparación: " + t["n_Reparación"].astype(str)
            + "<br>A atender: " + t["a_atender"].astype(str) + " de " + t["locales"].astype(str)
            + " (" + t["pct_a_atender"].astype(str) + " %)").tolist()

pri = {u: prioridad(u) for u in ubicaciones}
for u in ubicaciones:
    print(u, "por acción:", pri[u]["accion"].value_counts().reindex(lista_acciones, fill_value=0).to_dict())

# --- 2. Opciones del menú "Qué mostrar" ------------------------------
# (columna, título, escala, título de la barra, acción que se filtra)
opciones = [
    ("codigo",          "Acción principal",                                   escala_accion,           "Acción",     None),
    ("n_Conectividad",  "Conectividad (solo donde es la acción principal)",   ("#c7e9c0", colores["G3"]),  "Locales G3", "Conectividad"),
    ("n_Mantenimiento", "Mantenimiento (solo donde es la acción principal)",  ("#c6dbef", colores["G1"]),  "Locales G1", "Mantenimiento"),
    ("n_Reparación",    "Reparación (solo donde es la acción principal)",     ("#dadaeb", colores["G4"]),  "Locales G4", "Reparación"),
    ("pct_a_atender",   "% de locales a atender",                             "Reds",                  "% locales",  None),
]

def escala_con_gris(claro, oscuro, zmax):
    """0 = gris (otra acción); desde 1 local, degradado claro -> oscuro."""
    e = 0.5 / max(zmax, 1)
    return [[0, GRIS], [e, GRIS], [e, claro], [1, oscuro]]

# --- 3. Mapa base y capas de datos -----------------------------------
fig = go.Figure()
fig.add_trace(capa_base())
traza_opcion = [None]                      # None = traza fija (siempre visible)
trazas = {"Región": [], "Provincia": []}

for k, (columna, titulo, escala, titulo_barra, filtro) in enumerate(opciones):
    es_accion = columna == "codigo"
    for ubicacion in ubicaciones:
        t = pri[ubicacion]
        es_region = ubicacion == "Región"
        if filtro:   # se pinta solo donde la acción principal coincide; el resto = 0 (gris)
            z = t[columna].where(t["accion"] == filtro, 0)
            zmax = max(int(z.max()), 1)
            colorscale, zmin = escala_con_gris(*escala, zmax), 0
        elif es_accion:
            z, colorscale, zmin, zmax = t[columna], escala, -0.5, 3.5
        else:
            z, colorscale, zmin, zmax = t[columna], escala, 0, t[columna].max()
        fig.add_trace(go.Choropleth(
            geojson=geojson_de(ubicacion, t.index),
            featureidkey=f"properties.{ubicaciones[ubicacion]['clave']}",
            locations=t.index, z=z, colorscale=colorscale, zmin=zmin, zmax=zmax,
            text=texto_hover(t),
            hoverinfo="text" if es_region else "skip",
            marker=dict(opacity=0.85 if es_region else 0, line=dict(color="white", width=0.4)),
            showscale=es_region,
            colorbar=dict(title=titulo_barra, len=0.5, thickness=14, x=1.0, xanchor="left",
                          **(dict(tickvals=[0, 1, 2, 3], ticktext=lista_acciones) if es_accion else {})),
            visible=(k == 0)))
        trazas[ubicacion].append(len(fig.data) - 1)
        traza_opcion.append(k)

# --- 4. Capas de enfoque por departamento ----------------------------
# Límites (lon/lat) de cada departamento, calculados desde su geometría
def _puntos(c):
    if isinstance(c[0], (int, float)):
        yield c[:2]
    else:
        for s in c:
            yield from _puntos(s)

clave_reg = ubicaciones["Región"]["clave"]
geo_reg = geojson_de("Región", pri["Región"].index)
limites = {}
for f in geo_reg["features"]:
    p = np.array(list(_puntos(f["geometry"]["coordinates"])))
    limites[str(f["properties"][clave_reg])] = (p[:, 0].min(), p[:, 0].max(), p[:, 1].min(), p[:, 1].max())

codigos = list(pri["Región"].index)
TRANSPARENTE = "rgba(255,255,255,0)"

# 4a. Capa que atenúa los departamentos NO elegidos (al inicio, transparente)
fig.add_trace(go.Choropleth(
    geojson=geo_reg, featureidkey=f"properties.{clave_reg}",
    locations=codigos, z=[0] * len(codigos), zmin=0, zmax=1,
    colorscale=[[0, TRANSPARENTE], [1, "rgba(225,225,225,0.92)"]],
    showscale=False, hoverinfo="skip", marker=dict(line=dict(width=0))))
idx_dim = len(fig.data) - 1
traza_opcion.append(None)

# 4b. Contorno general de departamentos
fig.add_trace(capa_contorno())
traza_opcion.append(None)

# 4c. Borde grueso del departamento elegido (al inicio, vacío)
fig.add_trace(go.Choropleth(
    geojson=geo_reg, featureidkey=f"properties.{clave_reg}",
    locations=[], z=[], zmin=0, zmax=1,
    colorscale=[[0, TRANSPARENTE], [1, TRANSPARENTE]],
    showscale=False, hoverinfo="skip", marker=dict(line=dict(color="#111111", width=3))))
idx_borde = len(fig.data) - 1
traza_opcion.append(None)

# --- 5. Menú "Qué mostrar" ---------------------------------------------
botones = []
for k, (columna, titulo, *_r) in enumerate(opciones):
    visibles = [t is None or t == k for t in traza_opcion]
    botones.append(dict(label=titulo, method="update",
                        args=[{"visible": visibles}, {"title.text": titulo}]))
menu_variable = dict(buttons=botones, direction="down", x=0.20, y=1.09)

# --- 6. Menú "Enfocar departamento" ------------------------------------
# Todo el Perú: sin atenuar, sin borde, encuadre completo
botones_dep = [dict(label="Todo el Perú", method="update", args=[
    {"locations": [codigos, []], "z": [[0] * len(codigos), []]},
    {"geo.fitbounds": False, "geo.lonaxis.range": [-82, -68], "geo.lataxis.range": [-18.6, 0.2]},
    [idx_dim, idx_borde]])]

nombres_dep = pri["Región"]["nombre"].sort_values()
for cod, nombre in nombres_dep.items():
    otros = [c for c in codigos if c != cod]
    x0, x1, y0, y1 = limites[str(cod)]
    mx, my = (x1 - x0) * 0.10, (y1 - y0) * 0.10          # margen alrededor del departamento
    botones_dep.append(dict(label=nombre, method="update", args=[
        {"locations": [otros, [cod]], "z": [[1] * len(otros), [1]]},
        {"geo.fitbounds": False,
         "geo.lonaxis.range": [x0 - mx, x1 + mx],
         "geo.lataxis.range": [y0 - my, y1 + my]},
        [idx_dim, idx_borde]]))
menu_departamento = dict(buttons=botones_dep, direction="down", x=0.62, y=1.09)

# --- 7. Layout y menús ---------------------------------------------------
configurar_mapa(fig, opciones[0][1], [
    ("Ubicación", menu_ubicacion(trazas["Región"], trazas["Provincia"])),
    ("Variable", menu_variable),
    ("Departamento", menu_departamento)])

fig.update_layout(height=720, template="plotly_white", margin=dict(l=10, r=10, t=150, b=10),
                  font=dict(family="Arial", size=12),
                  title=dict(x=0.01, y=0.97, font=dict(size=18)),
                  hoverlabel=dict(bgcolor="white", font_size=12, align="left"))
fig.update_geos(fitbounds=False, visible=False, showcountries=False,
                showcoastlines=False, showland=False)

fig.layout.annotations = ()
for m, x, e in zip(fig.layout.updatemenus, [0.00, 0.20, 0.62],
                   ["Nivel", "Qué mostrar", "Enfocar departamento"]):
    m.update(x=x, y=1.09, xanchor="left", yanchor="top", bgcolor="white",
             bordercolor="#bbbbbb", font=dict(size=12))
    fig.add_annotation(text=f"<b>{e}</b>", xref="paper", yref="paper", x=x, y=1.14,
                       xanchor="left", yanchor="bottom", showarrow=False,
                       font=dict(size=11, color="#555555"))

fig.write_html("salidas/hito4_mapa_accion.html")
fig.show()

# --- 8. Ficha UGEL ---------------------------------------------------------
cuenta_ugel = pd.crosstab(locales_h4["cod_ugel"], locales_h4["perfil"])
ficha = locales_h4.groupby("cod_ugel").agg(
    ugel=("nombre_ugel", "first"),
    departamento=("dep_base", lambda s: s.value_counts().index[0]),
    locales=("perfil", "size"))
for perfil, accion in acciones.items():
    ficha[f"n_{accion}"] = cuenta_ugel[perfil]
ficha["a_atender"] = cuenta_ugel[list(acciones)].sum(axis=1)
ficha["pct_a_atender"] = (100 * ficha["a_atender"] / ficha["locales"]).round(1)
ficha["accion"] = accion_principal(cuenta_ugel)
ficha.sort_values("a_atender", ascending=False).to_csv("salidas/hito4_ficha_ugel.csv", encoding="utf-8-sig")
print("UGEL por acción:", ficha["accion"].value_counts().to_dict(), "| total:", len(ficha))
```

## **7. Problemas encontrados**

- Ward no se puede aplicar a los 42 215 locales por falta de memoria, por eso se usó una muestra de 5 000.
- DBSCAN sobre `Z_df` no encuentra regiones densas separadas: con `eps` bajo deja mucho ruido y con `eps` cercano al codo junta casi todo en un solo grupo.
- La cobertura móvil está a nivel distrital: todos los locales de un mismo distrito tienen el mismo valor, y es lo que más separa al grupo 0.
- 6 locales quedaron sin perfil por tener faltantes en la matriz analítica.
- El DBSCAN espacial usa un radio común de 10 km, elegido con G3: los porcentajes de aislados cambian con el radio, aunque el orden entre perfiles (G3 < G1 < G4) se mantiene entre 5 y 20 km.
- G0 no forma focos de brecha: sus focos son áreas metropolitanas continuas.
- Las zonas de los focos (5 km alrededor de cada local) de focos vecinos pueden tocarse en sus bordes, porque DBSCAN no encadena los locales del borde de un foco.
- Los focos no se han contrastado con la densidad del conjunto de locales.
